```python
#gerar o grafico, primeiro importar a biblioteca
import seaborn as sns
```

```python
#valores do grafico e definindo titulos para as colunas
grafico = sns.histplot(data=df,x="categoria",kde=True,bins=20)
grafico.set_title("Produtos vendidos")
grafico.set_xlabel("Produtos")
grafico.set_ylabel("Registros")
```

```python
#grafico do tipo boxplot(comparação)
grafico = sns.boxplot(data=df, x="categoria",y="preco_unitario")
grafico.set_title("Preco por categoria")
grafico.set_xlabel("Categorias")
grafico.set_ylabel("Preços")
```
